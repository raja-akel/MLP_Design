```python
import numpy as np
from functools import reduce
import operator
```

```python
class MLP_node:

    def __init__(self, n_inputs = 1, n_outputs = 1, weights = [1], biases = [0], activation_function = lambda x:1/(1 + np.exp(-x))):

        self.n_inputs = int(n_inputs)
        self.inputs = np.array((n_inputs, 1))

        self.n_outputs = int(n_outputs)
        self.outputs = np.array((n_outputs, 1))

        self.function = activation_function

        if len(weights) == n_inputs:
            self.weights = np.array(weights)
        elif len(weights) == 1:
            self.weights = np.array([weights] * n_inputs)
        else:
            print("Weights do not match the number of inputs")

        if len(biases) == n_inputs:
            self.biases = np.array(biases)
        elif len(biases) == 1:
            self.biases = np.array([biases] * n_inputs)
        else:
            print("Biases do not match the number of inputs")

    #setting
    def set_inputs(self, n_inputs):
        self.n_inputs = int(n_inputs)

    def set_outputs(self, n_outputs):
        self.n_outputs = int(n_outputs)

    def set_function(self, activation_function):
        self.function = activation_function

    def set_weights(self, weights):
        self.weights = np.array(weights)

    def give_input(self, input, index):
        self.inputs[index] = input

    #getting
    def get_parameters(self):
        return {
            "inputs" : self.n_inputs,
            "outputs" : self.n_outputs,
            "weights" : self.weights,
            "function" : self.function,
        }

    #behavior
    def run(self):
        intermediate_value =  reduce(operator.add , map(operator.mul, np.array(self.inputs), self.weights))
        return self.function(intermediate_value)
```

```python
class MLP_network:

    def __init__(self, n_inputs=1, n_outputs=1, n_layers=1, array_nodes=[1]):
        self.n_layers = int(n_layers)
        self.inputs = n_inputs
        self.outputs = n_outputs
        self.nodes_per_layer = list(array_nodes).append(self.outputs)
        self.depth = max(array_nodes)

        if len(array_nodes) == self.n_layers:

            #self.nodes = np.empty((self.depth, self.n_layers), dtype=MLP_node)
            self.nodes = np.full((self.depth, self.n_layers), None, dtype=object)

            for i in range(0, self.nodes.shape[1]): #for each layer
                for j in range(0, self.nodes_per_layer[i]): #for each node in layer
                    if i != 0:
                        self.nodes[j, i] = MLP_node(n_inputs=array_nodes[i-1])
                    else:
                        self.nodes[j, i] = MLP_node(n_inputs=self.inputs)

                    if i != self.nodes.shape[1] - 1:
                        self.nodes[j, i].set_outputs(array_nodes[i+1])
                    else:
                        self.nodes[j, i].set_outputs(self.outputs)
        else:
            print("number of layers does not match array given")

    def run(self, inputs):
        current_layer = 0
        current_depth = self.nodes_per_layer[current_layer]
        prev_array = np.array(inputs)
        next_array = np.array((self.nodes_per_layer[current_layer], 1))

        while current_layer < self.n_layers:
            for current_node in range(current_depth):
                for current_input in range(current_node.n_inputs):
                    self.nodes[current_node, current_layer].give_input(prev_array[current_input])


            current_layer += 1

        return next_array
```

```python
n_layers = 5
array_nodes = [3, 7, 4, 2, 5]
my_nodes = MLP_network(n_inputs=6, n_outputs=4, n_layers=n_layers, array_nodes=array_nodes)
```
