```python
import numpy as np
from functools import reduce
import operator
from random import random
```

```python
class MLP_Node:

    def __init__(self, weights = [1], biases = [0]):
        self.weights = np.array(weights).T
        self.biases = np.array(biases).T

    def set_weights(self, weights):
        self.weights = np.array(weights)

    def set_biases(self, biases):
        self.biases = np.array(biases)

    def get_params(self):
        print("W: ", self.weights)
        print ("B: ", self.biases)
```

```python
class MLP_Network:

    def __init__(self, nodes_per_layer = [3,3], n_inputs = 2, n_outputs=2):
        self.nodes = np.full((max(nodes_per_layer), len(nodes_per_layer)), None, dtype=object)
        self.layers = list([n_inputs] + list(nodes_per_layer) + [n_outputs])

    def randomize_nodes(self):
        for layer in range(self.nodes.shape[1]):
            for node in range(self.nodes.shape[0]):
                weights = np.array([
                    random()
                    for i in range(self.layers[layer]) #number of inputs
                ])
                biases = np.array([
                    random()
                    for i in range(self.layers[layer])
                ])
                self.nodes[node, layer] = MLP_Node(weights=weights, biases=biases)
        return

    def display_network(self):
        for layer in range(self.nodes.shape[1]):
            for node in range(self.nodes.shape[0]):
                self.nodes[node, layer].get_params()

    def run(self, inputs):
        current_layer = 1
        prev_values = np.array([inputs] * self.layers[current_layer])
        intermediate_values=[]
        next_values = np.array([[0] * self.layers[current_layer]] * self.layers[current_layer + 1])
        while current_layer < len(self.layers) - 1:
            intermediate_values = []
            for node_number in range(self.layers[current_layer]):
                I = prev_values[node_number]
                W = self.nodes[node_number, current_layer-1].weights
                B = self.nodes[node_number, current_layer-1].biases
                intermediate_values = np.dot(I, W)
                intermediate_values += B
                next_values[node_number] = np.array(intermediate_values)
            prev_values = next_values
            current_layer += 1
            next_values = np.array([[0] * self.layers[current_layer]] * self.layers[current_layer + 1]).T

        return prev_values
```

```python
my_MLP = MLP_Network()
my_MLP.randomize_nodes()
#my_MLP.display_network()

my_MLP.run([5, 2])
```
